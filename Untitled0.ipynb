## Resources

The following code demonstrates the use of SMOTE (Synthetic Minority Over-sampling Technique) to handle imbalanced datasets in binary classification. It includes:

1.  **Dataset Creation**: Generates an imbalanced synthetic dataset.
2.  **Baseline Model**: Trains a Logistic Regression model on the original imbalanced data.
3.  **SMOTE Application**: Applies SMOTE to the training data to balance the classes.
4.  **SMOTE Model**: Trains a Logistic Regression model on the SMOTE-balanced data.
5.  **Comparison**: Compares the performance of both models using classification reports, focusing on recall and F1-score for the minority class, and visualizes confusion matrices.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    f1_score,
    recall_score,
)

from imblearn.over_sampling import SMOTE
from collections import Counter

# ---------------------------------------------------------
# 1. Create an imbalanced dataset (e.g. 95% class 0, 5% class 1
#    -- like fraud detection, disease diagnosis, churn, etc.)
# ---------------------------------------------------------
X, y = make_classification(
    n_samples=2000,
    n_features=10,
    n_informative=5,
    n_redundant=2,
    n_clusters_per_class=1,
    weights=[0.95, 0.05],   # 95% majority, 5% minority
    flip_y=0.01,
    random_state=42,
)

print("Original class distribution:", Counter(y))

# ---------------------------------------------------------
# 2. Train/test split (stratified so both sets keep the same ratio)
# ---------------------------------------------------------
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, stratify=y, random_state=42
)

print("Train distribution:", Counter(y_train))
print("Test distribution:", Counter(y_test))

# ---------------------------------------------------------
# 3. BASELINE: train on the raw imbalanced training data
# ---------------------------------------------------------
baseline_model = LogisticRegression(max_iter=1000)
baseline_model.fit(X_train, y_train)
y_pred_baseline = baseline_model.predict(X_test)

print("\n================ BASELINE (no SMOTE) ================")
print(classification_report(y_test, y_pred_baseline, digits=3))

# ---------------------------------------------------------
# 4. Apply SMOTE ONLY to the training data
#    (never oversample the test set — that would leak info!)
# ---------------------------------------------------------
smote = SMOTE(random_state=42)
X_train_smote, y_train_smote = smote.fit_resample(X_train, y_train)

print("\nAfter SMOTE, train distribution:", Counter(y_train_smote))

# ---------------------------------------------------------
# 5. Train the SAME model type on the SMOTE-balanced data
# ---------------------------------------------------------
smote_model = LogisticRegression(max_iter=1000)
smote_model.fit(X_train_smote, y_train_smote)
y_pred_smote = smote_model.predict(X_test)

print("\n================ WITH SMOTE ================")
print(classification_report(y_test, y_pred_smote, digits=3))

# ---------------------------------------------------------
# 6. Side-by-side comparison of the key metrics for the
#    minority class (class 1) — this is usually what matters
# ---------------------------------------------------------
print("\n================ SUMMARY ================")
print(f"{'Metric':<25}{'Baseline':<12}{'With SMOTE':<12}")
print(f"{'Recall (class 1)':<25}"
      f"{recall_score(y_test, y_pred_baseline):<12.3f}"
      f"{recall_score(y_test, y_pred_smote):<12.3f}")
print(f"{'F1-score (class 1)':<25}"
      f"{f1_score(y_test, y_pred_baseline):<12.3f}"
      f"{f1_score(y_test, y_pred_smote):<12.3f}")

# ---------------------------------------------------------
# 7. Visualize confusion matrices side by side
# ---------------------------------------------------------
fig, axes = plt.subplots(1, 2, figsize=(11, 4.5))

ConfusionMatrixDisplay(
    confusion_matrix(y_test, y_pred_baseline)
).plot(ax=axes[0], colorbar=False)
axes[0].set_title("Baseline (no SMOTE)")

ConfusionMatrixDisplay(
    confusion_matrix(y_test, y_pred_smote)
).plot(ax=axes[1], colorbar=False)
axes[1].set_title("With SMOTE")

plt.tight_layout()
plt.savefig("confusion_matrices.png", dpi=150)
print("\nSaved confusion matrix comparison to 'confusion_matrices.png'")

# ---------------------------------------------------------
# Key takeaway:
# - Baseline models trained on imbalanced data often get high
#   overall accuracy just by predicting the majority class,
#   but RECALL on the minority class is usually poor.
# - SMOTE creates synthetic minority-class samples (by
#   interpolating between existing minority points and their
#   nearest neighbors) so the classifier sees a more balanced
#   training set, which typically improves minority-class recall
#   (often at some cost to precision — check the trade-off!).
# ---------------------------------------------------------
```
